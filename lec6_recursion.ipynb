## RECURSION:

Q: multiply a no. without using multiply operator.

```python
# FIRST, BY normal Function:

def mul(a,b):
    result=0

    for i in range(1,b+1):
        result=result+a

    print(result)

mul(5,7)        
```

```python
# BY RECURSION:
def recmul(a,b):
    if b==1:
        return a
    else:
        return a+ recmul(a,b-1)

recmul(5,6) 
     
```

fcatorial by recursion:

```python
def fac(n):
    if n==1 or n==0:
        return 1
    else:
        return n*fac(n-1) 

fac(5)
```

## THE RABBIT PROBLEM :

- a rabbit couple must be one month to reproduce , 
- will reproduce every month 
- one new offspring everymonth.
-never dies

```python
def fib(n):
    if n==0 or n==1:
        return 1
    else:
        return fib(n-1)+fib(n-2)

fib(48)
```
