## RECURSION:

Q: multiply a no. without using multiply operator.

```python
# FIRST, BY normal Function:

def mul(a,b):
    result=0

    for i in range(1,b+1):
        result=result+a

    print(result)

mul(5,7)        
```

```python
# BY RECURSION:
def recmul(a,b):
    if b==1:
        return a
    else:
        return a+ recmul(a,b-1)

recmul(5,6) 
     
```
