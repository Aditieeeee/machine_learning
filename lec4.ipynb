STRINGS ARE IMMUTABLE 👍
AS we can reassign the complete string . but cant ,modify or delete a particular index. 

```python
city='lol'
city[0]="i" # this will give error because string is immutable
```

```python
# deletion 
city='kolkata'
del city
city # this will give error because city is deleted 
```
