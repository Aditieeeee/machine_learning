dictionary code to get the maxm times which words has came in the lyrics

```python
lyrics="""Ho dil dhadkaye
Arrey Seeti bajaye
Haye dil dhadkaye
Seeti bajaye
Beech sadak pe nakhre dikhaye Saare
Oh..karke ishare
Ho ladki
aan,aan,aan,aan,aan,aan
Ho ladki
aan,aan,aan,aan,aan

Oh ladki aankh marey
Aankh marey
Oh ladki aankh marey
Aankh marey
Oh ladki aankh marey"""
p=lyrics.split()
print(p)
```

```python
# total words:
len(p)
```

```python
D={}
# a empty dict to store values
```

```python
for i in p:
    if i in D:
        D[i]=D[i]+1
    else:
        D[i]=1

print(D)            
```

```python
print(max(D.values()))

for i in D:
    if D[i]==5:
        print(i)
```
