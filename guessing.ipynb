GUESSING GAME:
>RANDOM NO. AND
 >USER HAS TO GUESS THE RANDOM NO. 
 >AND THE NO, OF MOVE SHE/SHE TAKES:

STEPS: WE IMPORT RANDOM
AND ATK S RANDOM NO,

```python
import random
num=random.randint(1,100)
c=0
guess=int(input("guess a no. from 1 to 100: "))
while guess!=num:
  if guess>num:
    print(" go lower")
    guess=int(input("guess again "))
    c+=1
  elif guess<num:
    print("go higher")
    guess=int(input("guess again "))
    c+=1
  else:
    print("not in range") 
else:
    print("you guessed it right")
    print(c)               
```
